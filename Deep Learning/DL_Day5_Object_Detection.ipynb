```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python
model=YOLO(model='yolo11n.pt')
```

```python
img=cv2.imread('image.jpg')
```

```python
plt.imshow(img)
plt.show()
```

```python
res=model(img)
```

```python
res
```

```python
annotated_img = res[0].plot()
plt.imshow(annotated_img)
plt.show()
```

```python
res=model(img,conf=0.5)
```

```python
annotated_img = res[0].plot()
plt.imshow(annotated_img)
plt.show()
```

```python
import os
out="./Object_detection/object_detection_output.jpeg"
os.makedirs(os.path.dirname(out), exist_ok=True)
cv2.imwrite(out, annotated_img)
```

```python
import cv2
import matplotlib.pyplot as plt

output_image_path = './Object_detection/object_detection_output.jpeg'
output_img = cv2.imread(output_image_path)

# OpenCV reads images in BGR, matplotlib expects RGB
output_img_rgb = cv2.cvtColor(output_img, cv2.COLOR_BGR2RGB)

plt.imshow(output_img_rgb)
plt.title('Saved Object Detection Output')
plt.axis('off') # Hide axes for cleaner display
plt.show()
```
